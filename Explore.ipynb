```python
import os
import glob
import json
import pandas as pd
from PIL import Image
from IPython.display import display, HTML
import ipyplot
```

```python
METADATA_PATH = 'metadata.csv'
YOLO_PREDICTIONS = "yolo_segmented/"
BIOCLIP_RESULT = 'bioclip-predictions.csv'
```

```python
metadata_df = pd.read_csv(METADATA_PATH)
metadata_df.head()
```

```python
bioclip_df = pd.read_csv(BIOCLIP_RESULT)
bioclip_df.head()
```

```python
image_path_to_yolo = {}
with open(os.path.join(YOLO_PREDICTIONS, "table_of_contents.json")) as infile:
    data = json.load(infile)
    for item in data:
        image_path_to_yolo[item['path']] = item
```

```python
def display_images(images, labels, img_width=300):
    if not images:
        print("No images found.")
    else:
        ipyplot.plot_class_representations(images, labels, img_width=img_width)
```

```python
def get_prediction(image_path):
    prediction = bioclip_df[bioclip_df['file_name'] == image_path]
    species = prediction['species'].values[0]
    common_name = prediction['common_name'].values[0]
    score = prediction['score'].values[0]
    return species, common_name, score
```

```python
for image_path, image_details in image_path_to_yolo.items():
    description = metadata_df[metadata_df['path'] == image_path]['description'].values
    species, common_name, score = get_prediction(image_path)
    print("#" * 100)    
    print(f"# Image {image_path} - {description}")
    print("#" * 100)
    print()
    
    print("Original")
    print(f"Prediction: {species} - {common_name} - {score}")
    display(Image.open(image_path).resize((400,400)))
    print()
    
    for segment_image_path in image_details['segments']:
        print(f"Segmented {segment_image_path}")
        display(Image.open(segment_image_path).resize((300,300)))
    print()
    
    for crop_image_path in image_details['crops']:
        print(f"Crop {crop_image_path}")
        species, common_name, score = get_prediction(crop_image_path)
        print(f"Prediction: {species} - {common_name} - {score}")        
        display(Image.open(crop_image_path).resize((300,300)))
    print()
    
    for mask_image_path in image_details['masks']:
        print(f"Mask {mask_image_path}")
        display(Image.open(mask_image_path).resize((200,200)))
    print()
    
    print()
```
